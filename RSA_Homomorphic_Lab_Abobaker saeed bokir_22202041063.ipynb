```python
# Choose small prime numbers
p = 5
q = 11

n = p * q
phi = (p - 1) * (q - 1)

print("p =", p)
print("q =", q)
print("n = p * q =", n)
print("phi(n) =", phi)
print("------------------------")

# Choose e (must be coprime with phi)
e = 3
print("e (public key) =", e)

# Find d such that (e * d) % phi == 1
d = None
for i in range(1, phi):
    if (e * i) % phi == 1:
        d = i
        break

print("d (private key) =", d)
print("------------------------")

# Basic power mod (beginner implementation)
def power_mod(base, exp, mod):
    result = 1
    for i in range(exp):
        result = (result * base) % mod
    return result

# Encryption and decryption
def encrypt(m, e, n):
    return power_mod(m, e, n)

def decrypt(c, d, n):
    return power_mod(c, d, n)

# Test messages
m1 = 6
m2 = 8

print("Original messages:")
print("m1 =", m1)
print("m2 =", m2)
print("------------------------")

# Encrypt messages
c1 = encrypt(m1, e, n)
c2 = encrypt(m2, e, n)

print("After encryption:")
print("E(m1) =", c1)
print("E(m2) =", c2)
print("------------------------")

# Decrypt to check correctness
dec1 = decrypt(c1, d, n)
dec2 = decrypt(c2, d, n)

print("Decryption check:")
print("D(E(m1)) =", dec1)
print("D(E(m2)) =", dec2)
print("------------------------")

# Show the homomorphic property E(m1)*E(m2) = E(m1*m2)
product_cipher = (c1 * c2) % n
print("E(m1) * E(m2) mod n =", product_cipher)

m_product = (m1 * m2) % n
enc_product = encrypt(m_product, e, n)

print("m1 * m2 mod n =", m_product)
print("E(m1 * m2) =", enc_product)
print("------------------------")

# Decrypt the product of ciphertexts
dec_product_cipher = decrypt(product_cipher, d, n)
print("Decrypting (E(m1) * E(m2)):")
print("D(E(m1) * E(m2)) =", dec_product_cipher)
print("------------------------")

# Final comparison
print("Does E(m1)*E(m2) mod n equal E(m1*m2)?")
print(product_cipher, "==", enc_product)
if product_cipher == enc_product:
    print("Result: YES, they are equal!")
else:
    print("Result: NO, they are not equal!")

print("Decrypted product =", dec_product_cipher)
print("Expected product =", m_product)
```
