```python
import hashlib
from ecdsa import SigningKey, SECP256k1

# Create a transaction message
transaction_message = "abobaker pays 100 coins to anas"
print("Transaction Message:", transaction_message)

# Generate an ECDSA private–public key pair
private_key = SigningKey.generate(curve=SECP256k1)   # ECDSA private key
public_key = private_key.get_verifying_key()         # ECDSA public key

print("\nPrivate Key (hex):", private_key.to_string().hex())
print("Public Key (hex):", public_key.to_string().hex())

# Hash the message using SHA-256
message_bytes = transaction_message.encode('utf-8')
message_hash = hashlib.sha256(message_bytes).digest()
print("\nSHA-256 Hash (hex):", message_hash.hex())

#  Sign the hash using the private key
signature = private_key.sign_digest(message_hash)
print("\nSignature (hex):", signature.hex())

#  Verify the signature using the public key
try:
    is_valid = public_key.verify_digest(signature, message_hash)
except Exception:
    is_valid = False

print("\nSignature Verified:", is_valid)
```
